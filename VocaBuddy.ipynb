```python
Word_Dict = {
    "transformer": {
        "chinese": "變形器",
        "correct": 0,
        "wrong": 0
    },
    "neural network": {
        "chinese": "神經網路",
        "correct": 0,
        "wrong": 0
    },
    "gradient descent": {
        "chinese": "梯度下降",
        "correct": 0,
        "wrong": 0
    },
    "overfitting": {
        "chinese": "過度擬合",
        "correct": 0,
        "wrong": 0
    },
    "underfitting": {
        "chinese": "欠擬合",
        "correct": 0,
        "wrong": 0
    },
    "backpropagation": {
        "chinese": "反向傳播",
        "correct": 0,
        "wrong": 0
    },
    "activation function": {
        "chinese": "激活函數",
        "correct": 0,
        "wrong": 0
    },
    "feature engineering": {
        "chinese": "特徵工程",
        "correct": 0,
        "wrong": 0
    },
    "deep learning": {
        "chinese": "深度學習",
        "correct": 0,
        "wrong": 0
    },
    "reinforcement learning": {
        "chinese": "強化學習",
        "correct": 0,
        "wrong": 0
    }
}
```

```python
def all_words():
    """取得所有單字。"""
    words = []
    for english, data in Word_Dict.items():
        words.append({
            "english": english,
            "chinese": data["chinese"],
            "correct": data["correct"],
            "wrong": data["wrong"]
        })
    return words
```

```python
def show_summary():
    words = all_words()

    total_correct = sum(word["correct"] for word in words)
    total_wrong = sum(word["wrong"] for word in words)

    print("=" * 40)
    print("             VocaBuddy")
    print("=" * 40)

    print(f"單字總數：{len(words)}")
    print(f"答對次數：{total_correct}")
    print(f"答錯次數：{total_wrong}")

    total = total_correct + total_wrong

    if total > 0:
        accuracy = total_correct / total * 100
        print(f"正確率：{accuracy:.1f}%")
    else:
        print("正確率：尚無紀錄")

    print("=" * 40)
```

```python
def show_word_progress():
    """顯示每個單字的熟練度。"""

    print("\n單字學習進度")
    print("-" * 40)

    for word in all_words():

        total = word["correct"] + word["wrong"]

        if total == 0:
            accuracy = 0
        else:
            accuracy = word["correct"] / total * 100

        print(
            f"{word['english']:25}"
            f"{word['chinese']:10}"
            f"正確 {word['correct']:2} 次  "
            f"錯誤 {word['wrong']:2} 次  "
            f"{accuracy:.0f}%"
        )
```

```python
def run_quiz(question_count=5):
    """進行單字測驗，並加入錯誤重測功能。"""

    candidates = all_words()

    if not candidates:
        print("找不到可測驗的單字。")
        return

    questions = random.sample(
        candidates,
        min(question_count, len(candidates))
    )

    correct_count = 0
    quiz_results = [] # To store results for later retesting

    print("\n===== 開始測驗 =====")

    for number, word in enumerate(questions, start=1):
        # Store the original word data for retesting
        quiz_results.append({"original_word": word, "answered_correctly": False})

        current_word_data = word
        attempt_count = 0

        while True:
            attempt_count += 1
            print(f"\n第 {number} 題 (嘗試 {attempt_count})")
            answer = input(f"{current_word_data['chinese']} 的英文是：").strip().lower()
            english = current_word_data["english"].lower()

            if answer == english:
                print("答對！")
                Word_Dict[current_word_data["english"]]["correct"] += 1
                quiz_results[number - 1]["answered_correctly"] = True # Mark as correct
                if attempt_count == 1: # Only count as correct if answered on first try for initial correct_count
                    correct_count += 1
                break  # Exit loop for this word, move to next question
            else:
                print(f"答錯！正確答案：{current_word_data['english']}")
                Word_Dict[current_word_data["english"]]["wrong"] += 1
                # For retesting, the loop continues for the same word until correct

    print("\n===== 測驗結果 =====")

    total_questions_asked = len(questions)
    print(f"答對（首次）：{correct_count}/{total_questions_asked}")

    # Recalculate accuracy based on first attempts, or overall for passing the quiz.
    # For this refactor, 'correct_count' reflects initial correct answers.
    accuracy = correct_count / total_questions_asked * 100
    print(f"正確率：{accuracy:.1f}%")
```

```python
def add_word(english, chinese):
    """新增單字，不再需要分類。"""

    english = english.strip().lower()
    chinese = chinese.strip()

    if english in Word_Dict:
        print(f"單字 '{english}' 已經存在。")
        return

    Word_Dict[english] = {
        "chinese": chinese,
        "correct": 0,
        "wrong": 0
    }

    print(f"已新增：{english} → {chinese}")


def delete_word(english):
    """刪除指定單字，不再需要分類。"""

    english = english.strip().lower()

    if english not in Word_Dict:
        print(f"找不到單字 '{english}'。")
        return

    del Word_Dict[english]

    print(f"已刪除：{english}")


def random_word():
    """隨機抽一個單字，不再需要分類。"""

    words = all_words()

    if not words:
        print("沒有單字可供抽取。")
        return

    word = random.choice(words)

    print("\n===== 隨機單字 =====")
    print(f"英文：{word['english']}")
    print(f"中文：{word['chinese']}")

```
